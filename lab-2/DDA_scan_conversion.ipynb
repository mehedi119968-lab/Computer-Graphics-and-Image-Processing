```python

import matplotlib.pyplot as plt



x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))

x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))



dx = x2 - x1
dy = y2 - y1



steps = max(abs(dx), abs(dy))



x_inc = dx / steps
y_inc = dy / steps


print("\nDDA Scan Conversion")
print("-------------------")

print("Delta X =", dx)
print("Delta Y =", dy)
print("Steps =", steps)

print("X Increment =", round(x_inc,3))
print("Y Increment =", round(y_inc,3))


# Generate DDA pixels

x = x1
y = y1

pixel_x = []
pixel_y = []


print("\nGenerated Pixels")
print("----------------")

for i in range(steps + 1):

    px = round(x)
    py = round(y)

    pixel_x.append(px)
    pixel_y.append(py)

    print(f"Pixel {i}: ({px},{py})")

    x += x_inc
    y += y_inc



# Plot

plt.figure(figsize=(8,6))


# Ideal mathematical line

plt.plot(
    [x1,x2],
    [y1,y2],
    color="blue",
    label="Ideal Mathematical Line"
)


# DDA generated pixels

plt.scatter(
    pixel_x,
    pixel_y,
    color="red",
    marker="s",
    s=100,
    label="DDA Generated Pixels"
)


# Start and end points

plt.scatter(
    [x1,x2],
    [y1,y2],
    color="blue",
    s=120,
    label="End Points"
)



# Pixel labels

for x,y in zip(pixel_x,pixel_y):

    plt.text(
        x+0.08,
        y+0.08,
        f"({x},{y})",
        fontsize=10
    )


plt.title("DDA Scan Conversion")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")


plt.grid(True)

plt.legend()

plt.axis("equal")

plt.show()
```
