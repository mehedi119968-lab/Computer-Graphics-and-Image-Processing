```python
import matplotlib.pyplot as plt


def line_scan_conversion(x1, y1, x2, y2):

    x_points = []
    y_points = []

    # Vertical line case
    if x1 == x2:
        print("\nVertical Line")

        step = 1 if y2 >= y1 else -1

        for y in range(y1, y2 + step, step):
            x_points.append(x1)
            y_points.append(y)
            print("X =", x1, "Y =", y)

        return x_points, y_points


    # Calculate slope and intercept
    m = (y2 - y1) / (x2 - x1)
    b = y1 - m * x1

    print("\nSlope =", round(m, 3))
    print("Intercept =", round(b, 3))


    # If slope <= 1, increment X
    if abs(m) <= 1:

        print("\nEquation: y = mx + b")
        print("X is incremented\n")

        step = 1 if x2 >= x1 else -1

        x = x1

        while True:

            y = m*x + b

            x_points.append(x)
            y_points.append(y)

            print(f"X = {x}, Y = {round(y,2)}")

            if x == x2:
                break

            x += step


    # If slope > 1, increment Y
    else:

        print("\nEquation: x = (y-b)/m")
        print("Y is incremented\n")

        step = 1 if y2 >= y1 else -1

        y = y1

        while True:

            x = (y-b)/m

            x_points.append(x)
            y_points.append(y)

            print(f"Y = {y}, X = {round(x,2)}")

            if y == y2:
                break

            y += step


    return x_points, y_points





x1 = int(input("Enter x1: "))
y1 = int(input("Enter y1: "))

x2 = int(input("Enter x2: "))
y2 = int(input("Enter y2: "))



x_values, y_values = line_scan_conversion(x1, y1, x2, y2)




plt.figure(figsize=(8,6))

plt.plot(
    x_values,
    y_values,
    'bo-',
    label="Scan Converted Points"
)

plt.scatter(
    [x1, x2],
    [y1, y2],
    color='red',
    s=80,
    label="End Points"
)

plt.xlabel("X-axis")
plt.ylabel("Y-axis")
plt.title("Straight Line Scan Conversion (DDA)")

plt.grid(True)
plt.legend()
plt.axis("equal")

plt.show()
```
