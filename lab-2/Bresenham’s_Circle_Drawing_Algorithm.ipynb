```python
# Bresenham's Circle Drawing Algorithm
# Circle Rasterization using Python

import matplotlib.pyplot as plt


def plot_circle_points(xc, yc, x, y, points):

    # 8-way symmetry of circle
    points.extend([
        (xc + x, yc + y),
        (xc - x, yc + y),
        (xc + x, yc - y),
        (xc - x, yc - y),

        (xc + y, yc + x),
        (xc - y, yc + x),
        (xc + y, yc - x),
        (xc - y, yc - x)
    ])



def bresenham_circle(xc, yc, radius):

    x = 0
    y = radius

    # Initial decision parameter
    decision = 3 - 2 * radius

    circle_points = []


    while x <= y:

        plot_circle_points(
            xc,
            yc,
            x,
            y,
            circle_points
        )


        if decision < 0:

            # Choose East pixel
            decision = decision + 4*x + 6

        else:

            # Choose South-East pixel
            decision = decision + 4*(x-y) + 10
            y = y - 1


        x = x + 1


    return circle_points



# -------- Input --------

xc = int(input("Enter center x-coordinate: "))
yc = int(input("Enter center y-coordinate: "))
r = int(input("Enter radius: "))


# Generate circle pixels
pixels = bresenham_circle(xc, yc, r)


print("\nGenerated Circle Pixels:")
print(pixels)



# Separate coordinates
x_values = [p[0] for p in pixels]
y_values = [p[1] for p in pixels]



# -------- Plot Circle --------

plt.figure(figsize=(7,7))

plt.scatter(
    x_values,
    y_values,
    s=50,
    label="Circle Pixels"
)


# Mark center
plt.scatter(
    xc,
    yc,
    color='red',
    s=100,
    label="Center"
)


plt.title("Bresenham's Circle Drawing Algorithm")

plt.xlabel("X-axis")
plt.ylabel("Y-axis")


plt.grid(True)
plt.legend()

plt.axis("equal")

plt.show()
```
