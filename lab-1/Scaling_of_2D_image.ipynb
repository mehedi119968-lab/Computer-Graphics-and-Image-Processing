```python
import cv2
import numpy as np
import matplotlib.pyplot as plt
```

```python
from google.colab import files

uploaded = files.upload()
```

```python
filename = next(iter(uploaded))

image = cv2.imread('/content/' + filename)

image = cv2.cvtColor(image, cv2.COLOR_BGR2RGB)

print("Image loaded successfully!")
```

```python
h, w = image.shape[:2]

print("Original Height:", h)
print("Original Width:", w)
```

```python
sx = 0.5
sy = 2.5

print("Scaling X:", sx)
print("Scaling Y:", sy)
```

```python
M = np.float32([
    [sx, 0, 0],
    [0, sy, 0]
])

print("Scaling Matrix:")
print(M)
```

```python
scaled = cv2.warpAffine(
    image,
    M,
    (int(w * sx), int(h * sy))
)
```

```python
plt.figure(figsize=(10, 5))

plt.subplot(1, 2, 1)
plt.imshow(image)
plt.title("Original Image")
plt.axis("off")

plt.subplot(1, 2, 2)
plt.imshow(scaled)
plt.title("Scaled Image")
plt.axis("off")

plt.show()
```
