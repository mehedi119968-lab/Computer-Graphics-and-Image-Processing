```python
import cv2
import numpy as np
import matplotlib.pyplot as plt

# Use the programmatically generated sample images instead of manual file upload
image_names = ['rectangle.png', 'circle.png']

if len(image_names) < 2:
    raise ValueError("Please upload two images!")

img1 = cv2.imread(image_names[0])
img2 = cv2.imread(image_names[1])

img2 = cv2.resize(img2, (img1.shape[1], img1.shape[0]))

and_result = cv2.bitwise_and(img1, img2)
or_result = cv2.bitwise_or(img1, img2)
not_result = cv2.bitwise_not(img1)

img1_rgb = cv2.cvtColor(img1, cv2.COLOR_BGR2RGB)
img2_rgb = cv2.cvtColor(img2, cv2.COLOR_BGR2RGB)
and_rgb = cv2.cvtColor(and_result, cv2.COLOR_BGR2RGB)
or_rgb = cv2.cvtColor(or_result, cv2.COLOR_BGR2RGB)
not_rgb = cv2.cvtColor(not_result, cv2.COLOR_BGR2RGB)

plt.figure(figsize=(15, 8))

plt.subplot(2, 3, 1)
plt.imshow(img1_rgb)
plt.title("Image 1 (Rectangle)")
plt.axis('off')

plt.subplot(2, 3, 2)
plt.imshow(img2_rgb)
plt.title("Image 2 (Circle)")
plt.axis('off')

plt.subplot(2, 3, 3)
plt.imshow(and_rgb)
plt.title("Bitwise AND")
plt.axis('off')

plt.subplot(2, 3, 4)
plt.imshow(or_rgb)
plt.title("Bitwise OR")
plt.axis('off')

plt.subplot(2, 3, 5)
plt.imshow(not_rgb)
plt.title("Bitwise NOT (Image 1)")
plt.axis('off')

plt.tight_layout()
plt.show()

cv2.imwrite("AND_Output.png", and_result)
cv2.imwrite("OR_Output.png", or_result)
cv2.imwrite("NOT_Output.png", not_result)

print("AND, OR and NOT Operations Completed Successfully!")
```

```python
import cv2
import numpy as np

# Create two black canvas images (300x300 pixels)
img1 = np.zeros((300, 300), dtype="uint8")
img2 = np.zeros((300, 300), dtype="uint8")

# Draw a white filled rectangle on the first image
cv2.rectangle(img1, (50, 50), (250, 250), 255, -1)

# Draw a white filled circle on the second image
cv2.circle(img2, (150, 150), 110, 255, -1)

# Save them to disk so they can be loaded by your script
cv2.imwrite('rectangle.png', img1)
cv2.imwrite('circle.png', img2)

print("Generated 'rectangle.png' and 'circle.png' successfully in your workspace!")
```
